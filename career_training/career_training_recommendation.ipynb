```python
# Separate input columns by data type

numeric_features = df[input_columns].select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_features = df[input_columns].select_dtypes(
    include=["object", "category"]
).columns.tolist()

print("NUMERIC FEATURES")
print("=" * 50)

for col in numeric_features:
    print(col)

print("\nTotal numeric features:", len(numeric_features))


print("\nCATEGORICAL FEATURES")
print("=" * 50)

for col in categorical_features:
    print(col)

print("\nTotal categorical features:", len(categorical_features))
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

# =========================================================
# PREPROCESSING
# =========================================================

preprocessor = ColumnTransformer(
    transformers=[

        # Numerical columns
        (
            "num",
            StandardScaler(),
            numeric_features
        ),

        # Categorical columns
        (
            "cat",
            OneHotEncoder(
                handle_unknown="ignore"
            ),
            categorical_features
        )
    ]
)

print("Preprocessor created successfully!")
```

```python
# =========================================================
# PREPARE TARGETS
# =========================================================

from sklearn.preprocessing import LabelEncoder

label_encoders = {}

for target in target_columns:

    le = LabelEncoder()

    df[target] = le.fit_transform(
        df[target].astype(str)
    )

    label_encoders[target] = le

print("Target encoding completed!")
```

```python
# =========================================================
# CREATE X AND Y
# =========================================================

X = df[input_columns]

Y = df[target_columns]

print("X shape:", X.shape)
print("Y shape:", Y.shape)

print("\nInput data:")
print(X.head())

print("\nTarget data:")
print(Y.head())
```

```python
# =========================================================
# TRAIN / TEST SPLIT
# =========================================================

from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(
    X,
    Y,
    test_size=0.20,
    random_state=42
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python
# =========================================================
# TRAIN RANDOM FOREST MODELS
# =========================================================

from sklearn.ensemble import RandomForestClassifier
from sklearn.multioutput import MultiOutputClassifier

# Create the Random Forest model
rf_model = RandomForestClassifier(
    n_estimators=300,
    max_depth=None,
    min_samples_split=2,
    min_samples_leaf=1,
    random_state=42,
    n_jobs=-1
)

# Handle the 5 career targets
model = MultiOutputClassifier(rf_model)

# Preprocess training data
X_train_processed = preprocessor.fit_transform(X_train)

print("Training model...")

# Train
model.fit(
    X_train_processed,
    Y_train
)

print("✅ Model training completed!")
```

```python
# =========================================================
# PREPARE TEST DATA
# =========================================================

X_test_processed = preprocessor.transform(X_test)

# Make predictions
Y_pred = model.predict(X_test_processed)

print("Prediction completed!")

print("Prediction shape:", Y_pred.shape)
print("Actual shape:", Y_test.shape)
```

```python
# =========================================================
# ACCURACY FOR EACH TOP CAREER
# =========================================================

from sklearn.metrics import accuracy_score

print("CAREER RECOMMENDATION ACCURACY")
print("=" * 50)

accuracies = {}

for i, target in enumerate(target_columns):

    accuracy = accuracy_score(
        Y_test.iloc[:, i],
        Y_pred[:, i]
    )

    accuracies[target] = accuracy

    print(
        f"{target}: {accuracy * 100:.2f}%"
    )
```

```python
# =========================================================
# STEP 6 - ENCODE CAREER TARGETS
# =========================================================

from sklearn.preprocessing import LabelEncoder

label_encoders = {}

for col in target_columns:

    encoder = LabelEncoder()

    y[col] = encoder.fit_transform(y[col].astype(str))

    label_encoders[col] = encoder

    print(f"\n{col}")
    print("Number of careers:", len(encoder.classes_))
    print("Example mapping:")

    for i, career in enumerate(encoder.classes_[:10]):
        print(i, "->", career)
```

```python
# =========================================================
# STEP 7 - TRAIN / TEST SPLIT
# =========================================================

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Data split completed!")
print("=" * 50)

print("Training samples:", len(X_train))
print("Testing samples :", len(X_test))

print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
```

```python
# =========================================================
# STEP 8 - PREPROCESS FEATURES
# =========================================================

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline

# Categorical column
categorical_features = ["field_filter"]

# Numerical columns
numerical_features = [
    col for col in feature_columns
    if col not in categorical_features
]

# Numerical preprocessing
numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median"))
])

# Complete preprocessor
preprocessor = ColumnTransformer(
    transformers=[
        (
            "categorical",
            OneHotEncoder(
                handle_unknown="ignore"
            ),
            categorical_features
        ),
        (
            "numerical",
            numeric_pipeline,
            numerical_features
        )
    ]
)

# Fit only on training data
X_train_processed = preprocessor.fit_transform(X_train)

# Transform test data
X_test_processed = preprocessor.transform(X_test)

print("Preprocessing completed!")
print("=" * 50)

print("Original training shape:", X_train.shape)
print("Processed training shape:", X_train_processed.shape)

print("Original testing shape:", X_test.shape)
print("Processed testing shape:", X_test_processed.shape)
```

```python
# =========================================================
# STEP 9 - TRAIN RANDOM FOREST MODEL
# =========================================================

from sklearn.ensemble import RandomForestClassifier
from sklearn.multioutput import MultiOutputClassifier

rf_base = RandomForestClassifier(
    n_estimators=500,
    max_depth=None,
    min_samples_split=2,
    min_samples_leaf=1,
    max_features="sqrt",
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)

rf_model = MultiOutputClassifier(
    rf_base,
    n_jobs=-1
)

print("Training Random Forest...")
print("Please wait...")

rf_model.fit(
    X_train_processed,
    y_train
)

print("✅ Random Forest training completed!")
```

```python
# =========================================================
# STEP 10 - EVALUATE RANDOM FOREST
# =========================================================

from sklearn.metrics import accuracy_score

rf_predictions = rf_model.predict(X_test_processed)

print("=" * 50)
print("RANDOM FOREST CAREER ACCURACY")
print("=" * 50)

accuracies = []

for i, col in enumerate(target_columns):

    accuracy = accuracy_score(
        y_test[col],
        rf_predictions[:, i]
    )

    accuracies.append(accuracy)

    print(f"{col}: {accuracy * 100:.2f}%")

average_accuracy = sum(accuracies) / len(accuracies)

print("=" * 50)
print(f"Average Accuracy: {average_accuracy * 100:.2f}%")
print("=" * 50)
```

```python
from sklearn.ensemble import HistGradientBoostingClassifier
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import OneHotEncoder
from sklearn.model_selection import train_test_split
import pandas as pd
import numpy as np


# =========================================================
# FEATURES
# =========================================================

feature_columns = [
    "field_filter",

    "hobby_coding",
    "hobby_gaming",
    "hobby_reading",
    "hobby_writing",
    "hobby_music",
    "hobby_drawing_art",
    "hobby_sports",
    "hobby_cooking",
    "hobby_photography",
    "hobby_travel",
    "hobby_science_experiments",
    "hobby_volunteering",
    "hobby_debating",
    "hobby_robotics",
    "hobby_fashion",
    "hobby_business_trading",

    "grade_math",
    "grade_science",
    "grade_lang",
    "grade_social",
    "grade_cs",

    "score_analytical",
    "score_numeric",
    "score_verbal",
    "score_creative",
    "score_social"
]


# =========================================================
# INPUT DATA
# =========================================================

X = df[feature_columns].copy()

X["field_filter"] = X["field_filter"].astype(str)


# =========================================================
# ENCODE FIELD
# =========================================================

field_encoder = OneHotEncoder(
    handle_unknown="ignore",
    sparse_output=False
)

X_field = field_encoder.fit_transform(
    X[["field_filter"]]
)

X_numeric = X.drop(
    columns=["field_filter"]
).values

X_processed = np.hstack([
    X_field,
    X_numeric
])


# =========================================================
# CAREER COLUMNS
# =========================================================

career_columns = [
    "top_career_1",
    "top_career_2",
    "top_career_3",
    "top_career_4",
    "top_career_5"
]


# =========================================================
# TRAIN 5 SEPARATE MODELS
# =========================================================

career_models = {}
career_predictions = {}

accuracies = []


for target_col in career_columns:

    print("=" * 60)
    print(target_col)
    print("=" * 60)

    y = df[target_col]

    # Remove missing values
    valid = y.notna()

    X_target = X_processed[valid]
    y_target = y[valid]


    # -----------------------------------------------------
    # TRAIN / TEST SPLIT
    # -----------------------------------------------------

    # IMPORTANT:
    # No stratify because some careers occur only once.

    X_train, X_test, y_train, y_test = train_test_split(
        X_target,
        y_target,
        test_size=0.20,
        random_state=42
    )


    # -----------------------------------------------------
    # MODEL
    # -----------------------------------------------------

    model = HistGradientBoostingClassifier(
        max_iter=300,
        learning_rate=0.08,
        max_leaf_nodes=31,
        l2_regularization=1.0,
        random_state=42
    )


    # -----------------------------------------------------
    # TRAIN
    # -----------------------------------------------------

    model.fit(
        X_train,
        y_train
    )


    # -----------------------------------------------------
    # PREDICT
    # -----------------------------------------------------

    pred = model.predict(X_test)


    # -----------------------------------------------------
    # ACCURACY
    # -----------------------------------------------------

    accuracy = accuracy_score(
        y_test,
        pred
    )

    accuracies.append(
        accuracy
    )


    # Save model
    career_models[target_col] = model

    career_predictions[target_col] = {
        "actual": y_test,
        "predicted": pred
    }


    print(
        f"{target_col}: "
        f"{accuracy * 100:.2f}%"
    )


# =========================================================
# FINAL RESULTS
# =========================================================

print()
print("=" * 60)
print("HISTOGRAM GRADIENT BOOSTING CAREER ACCURACY")
print("=" * 60)

for i, accuracy in enumerate(accuracies, 1):

    print(
        f"top_career_{i}: "
        f"{accuracy * 100:.2f}%"
    )


print("=" * 60)

average_accuracy = np.mean(accuracies)

print(
    f"Average Accuracy: "
    f"{average_accuracy * 100:.2f}%"
)

print("=" * 60)
```

```python
# ============================================================
# OPTIMIZED GRADIENT BOOSTING - COMPLETE SELF-CONTAINED CELL
# ============================================================

import pandas as pd
import numpy as np

from sklearn.ensemble import HistGradientBoostingClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import OneHotEncoder


# ============================================================
# 1. LOAD DATASET
# ============================================================

file_path = "career_training_data_cleaned (1).csv"

df = pd.read_csv(file_path)

print("=" * 60)
print("DATASET LOADED")
print("=" * 60)

print("Shape:", df.shape)


# ============================================================
# 2. CAREER TARGET COLUMNS
# ============================================================

career_columns = [
    "top_career_1",
    "top_career_2",
    "top_career_3",
    "top_career_4",
    "top_career_5"
]


# ============================================================
# 3. BASE FEATURES
# ============================================================

base_features = [

    "field_filter",

    "hobby_coding",
    "hobby_gaming",
    "hobby_reading",
    "hobby_writing",
    "hobby_music",
    "hobby_drawing_art",
    "hobby_sports",
    "hobby_cooking",
    "hobby_photography",
    "hobby_travel",
    "hobby_science_experiments",
    "hobby_volunteering",
    "hobby_debating",
    "hobby_robotics",
    "hobby_fashion",
    "hobby_business_trading",

    "grade_math",
    "grade_science",
    "grade_lang",
    "grade_social",
    "grade_cs",

    "score_analytical",
    "score_numeric",
    "score_verbal",
    "score_creative",
    "score_social"
]


# ============================================================
# 4. CREATE COPY
# ============================================================

df_model = df.copy()


# ============================================================
# 5. ACADEMIC FEATURES
# ============================================================

df_model["academic_average"] = (
    df_model["grade_math"]
    + df_model["grade_science"]
    + df_model["grade_lang"]
    + df_model["grade_social"]
    + df_model["grade_cs"]
) / 5


df_model["stem_average"] = (
    df_model["grade_math"]
    + df_model["grade_science"]
    + df_model["grade_cs"]
) / 3


df_model["language_social_average"] = (
    df_model["grade_lang"]
    + df_model["grade_social"]
) / 2


# ============================================================
# 6. APTITUDE FEATURES
# ============================================================

df_model["aptitude_average"] = (
    df_model["score_analytical"]
    + df_model["score_numeric"]
    + df_model["score_verbal"]
    + df_model["score_creative"]
    + df_model["score_social"]
) / 5


df_model["logical_strength"] = (
    df_model["score_analytical"]
    + df_model["score_numeric"]
) / 2


df_model["communication_strength"] = (
    df_model["score_verbal"]
    + df_model["score_social"]
) / 2


df_model["creative_strength"] = (
    df_model["score_creative"]
    + df_model["score_verbal"]
) / 2


# ============================================================
# 7. CAREER-RELATED FEATURES
# ============================================================

df_model["technology_strength"] = (
    df_model["grade_cs"]
    + df_model["score_analytical"]
    + df_model["score_numeric"]
    + df_model["hobby_coding"] * 100
    + df_model["hobby_robotics"] * 100
) / 5


df_model["science_strength"] = (
    df_model["grade_science"]
    + df_model["score_analytical"]
    + df_model["score_numeric"]
    + df_model["hobby_science_experiments"] * 100
) / 4


df_model["business_strength"] = (
    df_model["score_numeric"]
    + df_model["score_social"]
    + df_model["score_verbal"]
    + df_model["hobby_business_trading"] * 100
) / 4


df_model["creative_strength_combined"] = (
    df_model["score_creative"]
    + df_model["grade_lang"]
    + df_model["hobby_drawing_art"] * 100
    + df_model["hobby_music"] * 100
    + df_model["hobby_writing"] * 100
    + df_model["hobby_fashion"] * 100
) / 6


df_model["social_strength"] = (
    df_model["score_social"]
    + df_model["score_verbal"]
    + df_model["grade_social"]
    + df_model["hobby_debating"] * 100
    + df_model["hobby_volunteering"] * 100
) / 5


# ============================================================
# 8. HOBBY COUNT
# ============================================================

hobby_columns = [
    "hobby_coding",
    "hobby_gaming",
    "hobby_reading",
    "hobby_writing",
    "hobby_music",
    "hobby_drawing_art",
    "hobby_sports",
    "hobby_cooking",
    "hobby_photography",
    "hobby_travel",
    "hobby_science_experiments",
    "hobby_volunteering",
    "hobby_debating",
    "hobby_robotics",
    "hobby_fashion",
    "hobby_business_trading"
]


df_model["hobby_count"] = (
    df_model[hobby_columns].sum(axis=1)
)


# ============================================================
# 9. OVERALL STRENGTH
# ============================================================

df_model["overall_strength"] = (
    df_model["academic_average"]
    + df_model["aptitude_average"]
) / 2


# ============================================================
# 10. ALL FEATURES
# ============================================================

new_features = [

    "academic_average",
    "stem_average",
    "language_social_average",
    "aptitude_average",
    "logical_strength",
    "communication_strength",
    "creative_strength",
    "technology_strength",
    "science_strength",
    "business_strength",
    "creative_strength_combined",
    "social_strength",
    "hobby_count",
    "overall_strength"
]


feature_columns = base_features + new_features


print()
print("=" * 60)
print("FEATURE ENGINEERING COMPLETE")
print("=" * 60)

print("Total features:", len(feature_columns))


# ============================================================
# 11. PREPARE X
# ============================================================

X = df_model[feature_columns].copy()

X["field_filter"] = X["field_filter"].astype(str)


# ============================================================
# 12. ONE-HOT ENCODE FIELD
# ============================================================

field_encoder = OneHotEncoder(
    handle_unknown="ignore",
    sparse_output=False
)


X_field = field_encoder.fit_transform(
    X[["field_filter"]]
)


# ============================================================
# 13. NUMERIC FEATURES
# ============================================================

X_numeric = X.drop(
    columns=["field_filter"]
).values


# ============================================================
# 14. COMBINE FEATURES
# ============================================================

X_processed = np.hstack([
    X_field,
    X_numeric
])


print("Processed feature shape:", X_processed.shape)


# ============================================================
# 15. MODEL SETTINGS
# ============================================================

settings = [

    {
        "max_iter": 500,
        "learning_rate": 0.04,
        "max_leaf_nodes": 31,
        "min_samples_leaf": 5,
        "l2_regularization": 1.0
    },

    {
        "max_iter": 600,
        "learning_rate": 0.03,
        "max_leaf_nodes": 63,
        "min_samples_leaf": 5,
        "l2_regularization": 1.0
    },

    {
        "max_iter": 700,
        "learning_rate": 0.03,
        "max_leaf_nodes": 31,
        "min_samples_leaf": 3,
        "l2_regularization": 2.0
    }
]


# ============================================================
# 16. TRAIN MODELS
# ============================================================

tuned_models = {}
tuned_accuracies = {}


for target_col in career_columns:

    print()
    print("=" * 60)
    print(target_col)
    print("=" * 60)

    y = df_model[target_col]

    # Remove missing target values
    valid = y.notna()

    X_target = X_processed[valid]
    y_target = y[valid]


    # --------------------------------------------------------
    # Remove classes with only ONE sample
    # --------------------------------------------------------

    class_counts = y_target.value_counts()

    valid_classes = class_counts[
        class_counts >= 2
    ].index

    keep = y_target.isin(valid_classes)

    X_target = X_target[keep]
    y_target = y_target[keep]


    print("Samples:", len(y_target))
    print("Classes:", y_target.nunique())


    # --------------------------------------------------------
    # Train/Test Split
    # --------------------------------------------------------

    X_train, X_test, y_train, y_test = train_test_split(

        X_target,
        y_target,

        test_size=0.20,

        random_state=42,

        stratify=y_target
    )


    best_accuracy = 0
    best_model = None
    best_setting = None


    # --------------------------------------------------------
    # TEST DIFFERENT SETTINGS
    # --------------------------------------------------------

    for number, params in enumerate(settings, 1):

        print(
            f"Testing configuration "
            f"{number}/{len(settings)}..."
        )


        model = HistGradientBoostingClassifier(

            max_iter=params["max_iter"],

            learning_rate=params["learning_rate"],

            max_leaf_nodes=params["max_leaf_nodes"],

            min_samples_leaf=params["min_samples_leaf"],

            l2_regularization=params["l2_regularization"],

            early_stopping=True,

            random_state=42
        )


        model.fit(
            X_train,
            y_train
        )


        pred = model.predict(
            X_test
        )


        accuracy = accuracy_score(
            y_test,
            pred
        )


        print(
            f"  Accuracy: "
            f"{accuracy * 100:.2f}%"
        )


        # Keep best
        if accuracy > best_accuracy:

            best_accuracy = accuracy

            best_model = model

            best_setting = number


    # --------------------------------------------------------
    # SAVE BEST
    # --------------------------------------------------------

    tuned_models[target_col] = best_model

    tuned_accuracies[target_col] = best_accuracy


    print()
    print(
        f"BEST FOR {target_col}: "
        f"{best_accuracy * 100:.2f}%"
    )

    print(
        f"Configuration: {best_setting}"
    )


# ============================================================
# 17. FINAL RESULTS
# ============================================================

print()
print("=" * 60)
print("TUNED GRADIENT BOOSTING RESULTS")
print("=" * 60)


for target_col in career_columns:

    accuracy = tuned_accuracies[target_col]

    print(
        f"{target_col}: "
        f"{accuracy * 100:.2f}%"
    )


average_accuracy = np.mean(
    list(tuned_accuracies.values())
)


print("=" * 60)

print(
    f"Average Accuracy: "
    f"{average_accuracy * 100:.2f}%"
)

print("=" * 60)


# ============================================================
# 18. OLD VS NEW
# ============================================================

old_average = 76.41

new_average = average_accuracy * 100

print()
print("=" * 60)
print("OLD VS NEW")
print("=" * 60)

print(
    f"Old average: "
    f"{old_average:.2f}%"
)

print(
    f"New average: "
    f"{new_average:.2f}%"
)

print(
    f"Improvement: "
    f"{new_average - old_average:+.2f}%"
)

print("=" * 60)
```

```python
%%writefile app.py

import streamlit as st
import pandas as pd
import numpy as np
import joblib
from pathlib import Path


# ============================================================
# PAGE SETTINGS
# ============================================================

st.set_page_config(
    page_title="Career Recommendation System",
    page_icon="🎓",
    layout="wide"
)


# ============================================================
# LOAD MODEL
# ============================================================

@st.cache_resource
def load_model():

    model_path = Path(__file__).parent / "career_ranking_model.pkl"

    if not model_path.exists():

        st.error(
            "❌ career_ranking_model.pkl not found."
        )

        st.stop()

    return joblib.load(model_path)


data = load_model()

ranking_model = data["model"]
feature_columns = data["feature_columns"]


# ============================================================
# TITLE
# ============================================================

st.title("🎓 Career Recommendation System")

st.write(
    "Enter your field, hobbies, academic grades and aptitude "
    "scores to receive your Top 5 career recommendations."
)


# ============================================================
# FIELD
# ============================================================

st.subheader("🎓 Select Your Field")

fields = [
    "Engineering",
    "Medical & Healthcare",
    "Business & Finance",
    "Science & Research",
    "Arts & Design",
    "Computer Science & IT",
    "Education",
    "Law & Government",
    "Media & Communication",
    "Sports & Fitness"
]

field = st.selectbox(
    "Choose your field",
    fields
)


# ============================================================
# STUDENT DATA
# ============================================================

student = {}


# ============================================================
# HOBBIES
# ============================================================

st.subheader("🎯 Hobbies")

hobby_columns = [

    "hobby_coding",
    "hobby_gaming",
    "hobby_reading",
    "hobby_writing",
    "hobby_music",
    "hobby_drawing_art",
    "hobby_sports",
    "hobby_cooking",
    "hobby_photography",
    "hobby_travel",
    "hobby_science_experiments",
    "hobby_volunteering",
    "hobby_debating",
    "hobby_robotics",
    "hobby_fashion",
    "hobby_business_trading"

]


cols = st.columns(4)


for i, hobby in enumerate(hobby_columns):

    name = (
        hobby
        .replace("hobby_", "")
        .replace("_", " ")
        .title()
    )

    with cols[i % 4]:

        selected = st.checkbox(name)

    student[hobby] = 1 if selected else 0


# ============================================================
# ACADEMIC GRADES
# ============================================================

st.subheader("📚 Academic Grades")

grade_columns = [

    "grade_math",
    "grade_science",
    "grade_lang",
    "grade_social",
    "grade_cs"

]


grade_names = {

    "grade_math": "Math",
    "grade_science": "Science",
    "grade_lang": "Language",
    "grade_social": "Social Science",
    "grade_cs": "Computer Science"

}


cols = st.columns(5)


for i, grade in enumerate(grade_columns):

    with cols[i]:

        student[grade] = st.number_input(

            grade_names[grade],

            min_value=0.0,
            max_value=100.0,
            value=50.0,
            step=1.0

        )


# ============================================================
# APTITUDE SCORES
# ============================================================

st.subheader("🧠 Aptitude Scores")

score_columns = [

    "score_analytical",
    "score_numeric",
    "score_verbal",
    "score_creative",
    "score_social"

]


score_names = {

    "score_analytical": "Analytical",
    "score_numeric": "Numeric",
    "score_verbal": "Verbal",
    "score_creative": "Creative",
    "score_social": "Social"

}


cols = st.columns(5)


for i, score in enumerate(score_columns):

    with cols[i]:

        student[score] = st.number_input(

            score_names[score],

            min_value=0.0,
            max_value=100.0,
            value=50.0,
            step=1.0

        )


# ============================================================
# FEATURE ENGINEERING
# ============================================================

def create_features(student):

    df = pd.DataFrame([student])


    # --------------------------------------------------------
    # Academic
    # --------------------------------------------------------

    df["academic_average"] = (

        df["grade_math"]
        + df["grade_science"]
        + df["grade_lang"]
        + df["grade_social"]
        + df["grade_cs"]

    ) / 5


    df["stem_average"] = (

        df["grade_math"]
        + df["grade_science"]
        + df["grade_cs"]

    ) / 3


    df["language_social_average"] = (

        df["grade_lang"]
        + df["grade_social"]

    ) / 2


    # --------------------------------------------------------
    # Aptitude
    # --------------------------------------------------------

    df["aptitude_average"] = (

        df["score_analytical"]
        + df["score_numeric"]
        + df["score_verbal"]
        + df["score_creative"]
        + df["score_social"]

    ) / 5


    df["logical_strength"] = (

        df["score_analytical"]
        + df["score_numeric"]

    ) / 2


    df["communication_strength"] = (

        df["score_verbal"]
        + df["score_social"]

    ) / 2


    df["creative_strength"] = (

        df["score_creative"]
        + df["score_verbal"]

    ) / 2


    # --------------------------------------------------------
    # Career-related features
    # --------------------------------------------------------

    df["technology_strength"] = (

        df["grade_cs"]
        + df["score_analytical"]
        + df["score_numeric"]
        + df["hobby_coding"] * 100
        + df["hobby_robotics"] * 100

    ) / 5


    df["science_strength"] = (

        df["grade_science"]
        + df["score_analytical"]
        + df["score_numeric"]
        + df["hobby_science_experiments"] * 100

    ) / 4


    df["business_strength"] = (

        df["score_numeric"]
        + df["score_social"]
        + df["score_verbal"]
        + df["hobby_business_trading"] * 100

    ) / 4


    df["creative_strength_combined"] = (

        df["score_creative"]
        + df["grade_lang"]
        + df["hobby_drawing_art"] * 100
        + df["hobby_music"] * 100
        + df["hobby_writing"] * 100
        + df["hobby_fashion"] * 100

    ) / 6


    df["social_strength"] = (

        df["score_social"]
        + df["score_verbal"]
        + df["grade_social"]
        + df["hobby_debating"] * 100
        + df["hobby_volunteering"] * 100

    ) / 5


    # --------------------------------------------------------
    # Hobby count
    # --------------------------------------------------------

    df["hobby_count"] = df[hobby_columns].sum(axis=1)


    # --------------------------------------------------------
    # Overall strength
    # --------------------------------------------------------

    df["overall_strength"] = (

        df["academic_average"]
        + df["aptitude_average"]

    ) / 2


    return df


# ============================================================
# FIELD-BASED CAREER BOOST
# ============================================================
#
# The current ranking model was trained with 40 numerical
# features and does NOT contain field_filter.
#
# Therefore, we apply a small field compatibility boost here.
#
# This makes changing the selected field actually influence
# the final ranking.
# ============================================================

field_keywords = {

    "Engineering": [

        "engineer",
        "architect",
        "technician"

    ],

    "Medical & Healthcare": [

        "doctor",
        "nurse",
        "medical",
        "physiotherapist",
        "psychologist",
        "health",
        "laboratory"

    ],

    "Business & Finance": [

        "accountant",
        "manager",
        "marketing",
        "business",
        "finance",
        "financial",
        "trading",
        "operations",
        "human resource",
        "hr",
        "economist"

    ],

    "Science & Research": [

        "scientist",
        "research",
        "physicist",
        "chemist",
        "biologist",
        "astronomer",
        "mathematician",
        "statistician",
        "geologist",
        "oceanographer",
        "ocean",
        "space"

    ],

    "Arts & Design": [

        "designer",
        "artist",
        "animator",
        "fashion",
        "interior",
        "vfx",
        "illustrator",
        "drawing"

    ],

    "Computer Science & IT": [

        "software",
        "developer",
        "programmer",
        "data scientist",
        "data analyst",
        "machine learning",
        "ai engineer",
        "cloud engineer",
        "devops",
        "network engineer",
        "cyber",
        "forensics",
        "web developer",
        "game developer",
        "it project"

    ],

    "Education": [

        "teacher",
        "professor",
        "trainer",
        "education",
        "special education",
        "instructor"

    ],

    "Law & Government": [

        "lawyer",
        "advocate",
        "government",
        "officer",
        "naval",
        "coast guard",
        "crime",
        "forensic"

    ],

    "Media & Communication": [

        "journalist",
        "writer",
        "content",
        "editor",
        "film",
        "director",
        "media",
        "public relations",
        "social media",
        "video"

    ],

    "Sports & Fitness": [

        "coach",
        "fitness",
        "sports",
        "physiotherapist",
        "yoga",
        "athlete"

    ]

}


def apply_field_boost(results, selected_field):

    keywords = field_keywords.get(
        selected_field,
        []
    )

    def calculate_boost(career):

        career_text = str(career).lower()

        for keyword in keywords:

            if keyword.lower() in career_text:

                return 0.20

        return 0.0


    results["Field_Boost"] = results["Career"].apply(
        calculate_boost
    )


    results["Final_Score"] = (

        results["Score"]
        + results["Field_Boost"]

    )


    return results


# ============================================================
# PREDICTION
# ============================================================

st.divider()


if st.button(

    "🔮 Predict Top 5 Careers",

    use_container_width=True,

    type="primary"

):

    try:

        # ----------------------------------------------------
        # CREATE FEATURES
        # ----------------------------------------------------

        student_df = create_features(student)


        # ----------------------------------------------------
        # CHECK FEATURES
        # ----------------------------------------------------

        missing_features = [

            col

            for col in feature_columns

            if col not in student_df.columns

        ]


        if missing_features:

            st.error(
                "❌ Some model features are missing."
            )

            st.write(missing_features)

            st.stop()


        # ----------------------------------------------------
        # EXACT FEATURE ORDER
        # ----------------------------------------------------

        X = student_df[
            feature_columns
        ].copy()


        X_processed = X.values


        # ----------------------------------------------------
        # MODEL PROBABILITIES
        # ----------------------------------------------------

        probabilities = ranking_model.predict_proba(
            X_processed
        )[0]


        classes = ranking_model.classes_


        # ----------------------------------------------------
        # CREATE RESULTS
        # ----------------------------------------------------

        results = pd.DataFrame({

            "Career": classes,

            "Score": probabilities

        })


        # ====================================================
        # REMOVE NOT_RECOMMENDED
        # ====================================================

        results = results[

            results["Career"].astype(str).str.upper()
            != "NOT_RECOMMENDED"

        ].copy()


        # ----------------------------------------------------
        # NORMALIZE MODEL SCORES
        # ----------------------------------------------------

        total_score = results["Score"].sum()


        if total_score > 0:

            results["Score"] = (
                results["Score"]
                / total_score
            )


        # ----------------------------------------------------
        # APPLY FIELD BOOST
        # ----------------------------------------------------

        results = apply_field_boost(
            results,
            field
        )


        # ----------------------------------------------------
        # SORT FINAL RESULTS
        # ----------------------------------------------------

        results = results.sort_values(

            by="Final_Score",

            ascending=False

        ).reset_index(drop=True)


        # ----------------------------------------------------
        # NORMALIZE FINAL SCORES
        # ----------------------------------------------------

        total_final = results["Final_Score"].sum()


        if total_final > 0:

            results["Final_Score"] = (

                results["Final_Score"]
                / total_final

            )


        # ----------------------------------------------------
        # TOP 5
        # ----------------------------------------------------

        top_5 = results.head(5).copy()

        top_5 = top_5.reset_index(drop=True)


        # ====================================================
        # DISPLAY
        # ====================================================

        st.subheader(
            "🏆 Your Top 5 Career Recommendations"
        )


        st.write(

            f"Recommendations are based on your "
            f"**{field}** field, hobbies, grades and "
            f"aptitude scores."

        )


        # ====================================================
        # CAREER CARDS
        # ====================================================

        for i in range(len(top_5)):

            career = top_5.loc[
                i,
                "Career"
            ]

            score = top_5.loc[
                i,
                "Final_Score"
            ] * 100


            if i == 0:

                st.success(

                    f"🥇 **1. {career}**\n\n"
                    f"Suitability Score: **{score:.2f}%**"

                )


            elif i == 1:

                st.info(

                    f"🥈 **2. {career}**\n\n"
                    f"Suitability Score: **{score:.2f}%**"

                )


            elif i == 2:

                st.warning(

                    f"🥉 **3. {career}**\n\n"
                    f"Suitability Score: **{score:.2f}%**"

                )


            else:

                st.write(
                    f"### {i + 1}. {career}"
                )


                st.progress(

                    min(
                        float(
                            top_5.loc[
                                i,
                                "Final_Score"
                            ]
                        ),
                        1.0
                    )

                )


                st.write(

                    f"Suitability Score: "
                    f"**{score:.2f}%**"

                )


        # ====================================================
        # SUMMARY TABLE
        # ====================================================

        st.divider()

        st.subheader(
            "📊 Ranking Summary"
        )


        display_results = top_5[

            [
                "Career",
                "Final_Score"
            ]

        ].copy()


        display_results.insert(

            0,

            "Rank",

            range(
                1,
                len(display_results) + 1
            )

        )


        display_results["Final_Score"] = (

            display_results["Final_Score"]
            * 100

        ).round(2)


        display_results = (
            display_results
            .rename(
                columns={
                    "Final_Score":
                    "Suitability (%)"
                }
            )
        )


        st.dataframe(

            display_results,

            use_container_width=True,

            hide_index=True

        )


        # ====================================================
        # DEBUG / INFORMATION
        # ====================================================

        with st.expander(
            "🔍 See Recommendation Details"
        ):

            st.write(
                "Selected Field:",
                field
            )

            st.write(
                "Number of careers evaluated:",
                len(results)
            )

            st.write(
                "NOT_RECOMMENDED removed:",
                "Yes"
            )


    except Exception as e:

        st.error(
            "❌ Prediction failed."
        )

        st.code(
            str(e)
        )
```

```python
# ============================================================
# IMPROVED CAREER RANKING MODEL
# ============================================================

import pandas as pd
import numpy as np
import joblib
import os

from sklearn.ensemble import HistGradientBoostingClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score


print("=" * 60)
print("BUILDING IMPROVED CAREER RANKING MODEL")
print("=" * 60)


# ============================================================
# 1. LOAD DATASET
# ============================================================

csv_files = [
    f for f in os.listdir(".")
    if f.endswith(".csv")
]

print("CSV files found:", csv_files)

if "career_training_data_cleaned.csv" in csv_files:
    csv_file = "career_training_data_cleaned.csv"
else:
    cleaned_files = [
        f for f in csv_files
        if f.startswith("career_training_data_cleaned")
    ]

    if not cleaned_files:
        raise FileNotFoundError(
            "Career CSV dataset not found."
        )

    csv_file = cleaned_files[0]

df = pd.read_csv(csv_file)

print("Dataset:", csv_file)
print("Shape:", df.shape)


# ============================================================
# 2. CAREER COLUMNS
# ============================================================

career_columns = [
    "top_career_1",
    "top_career_2",
    "top_career_3",
    "top_career_4",
    "top_career_5"
]


# ============================================================
# 3. FEATURE COLUMNS
# ============================================================

hobby_columns = [
    "hobby_coding",
    "hobby_gaming",
    "hobby_reading",
    "hobby_writing",
    "hobby_music",
    "hobby_drawing_art",
    "hobby_sports",
    "hobby_cooking",
    "hobby_photography",
    "hobby_travel",
    "hobby_science_experiments",
    "hobby_volunteering",
    "hobby_debating",
    "hobby_robotics",
    "hobby_fashion",
    "hobby_business_trading"
]


grade_columns = [
    "grade_math",
    "grade_science",
    "grade_lang",
    "grade_social",
    "grade_cs"
]


score_columns = [
    "score_analytical",
    "score_numeric",
    "score_verbal",
    "score_creative",
    "score_social"
]


# ============================================================
# 4. FEATURE ENGINEERING
# ============================================================

df_model = df.copy()


# ------------------------------------------------------------
# Academic features
# ------------------------------------------------------------

df_model["academic_average"] = (
    df_model["grade_math"]
    + df_model["grade_science"]
    + df_model["grade_lang"]
    + df_model["grade_social"]
    + df_model["grade_cs"]
) / 5


df_model["stem_average"] = (
    df_model["grade_math"]
    + df_model["grade_science"]
    + df_model["grade_cs"]
) / 3


df_model["language_social_average"] = (
    df_model["grade_lang"]
    + df_model["grade_social"]
) / 2


# ------------------------------------------------------------
# Aptitude features
# ------------------------------------------------------------

df_model["aptitude_average"] = (
    df_model["score_analytical"]
    + df_model["score_numeric"]
    + df_model["score_verbal"]
    + df_model["score_creative"]
    + df_model["score_social"]
) / 5


df_model["logical_strength"] = (
    df_model["score_analytical"]
    + df_model["score_numeric"]
) / 2


df_model["communication_strength"] = (
    df_model["score_verbal"]
    + df_model["score_social"]
) / 2


df_model["creative_strength"] = (
    df_model["score_creative"]
    + df_model["score_verbal"]
) / 2


# ------------------------------------------------------------
# Career-related features
# ------------------------------------------------------------

df_model["technology_strength"] = (
    df_model["grade_cs"]
    + df_model["score_analytical"]
    + df_model["score_numeric"]
    + df_model["hobby_coding"] * 100
    + df_model["hobby_robotics"] * 100
) / 5


df_model["science_strength"] = (
    df_model["grade_science"]
    + df_model["score_analytical"]
    + df_model["score_numeric"]
    + df_model["hobby_science_experiments"] * 100
) / 4


df_model["business_strength"] = (
    df_model["score_numeric"]
    + df_model["score_social"]
    + df_model["score_verbal"]
    + df_model["hobby_business_trading"] * 100
) / 4


df_model["creative_strength_combined"] = (
    df_model["score_creative"]
    + df_model["grade_lang"]
    + df_model["hobby_drawing_art"] * 100
    + df_model["hobby_music"] * 100
    + df_model["hobby_writing"] * 100
    + df_model["hobby_fashion"] * 100
) / 6


df_model["social_strength"] = (
    df_model["score_social"]
    + df_model["score_verbal"]
    + df_model["grade_social"]
    + df_model["hobby_debating"] * 100
    + df_model["hobby_volunteering"] * 100
) / 5


# ------------------------------------------------------------
# Hobby count
# ------------------------------------------------------------

df_model["hobby_count"] = (
    df_model[hobby_columns].sum(axis=1)
)


# ------------------------------------------------------------
# Overall strength
# ------------------------------------------------------------

df_model["overall_strength"] = (
    df_model["academic_average"]
    + df_model["aptitude_average"]
) / 2


# ============================================================
# 5. RANKING FEATURES
# ============================================================

ranking_features = (
    hobby_columns
    + grade_columns
    + score_columns
    + [
        "academic_average",
        "stem_average",
        "language_social_average",
        "aptitude_average",
        "logical_strength",
        "communication_strength",
        "creative_strength",
        "technology_strength",
        "science_strength",
        "business_strength",
        "creative_strength_combined",
        "social_strength",
        "hobby_count",
        "overall_strength"
    ]
)


X = df_model[ranking_features].copy()


print()
print("=" * 60)
print("FEATURE ENGINEERING COMPLETE")
print("=" * 60)

print("Total ranking features:", len(ranking_features))
print("Processed shape:", X.shape)


# ============================================================
# 6. CREATE CAREER SCORES
# ============================================================

print()
print("=" * 60)
print("CREATING CAREER SUITABILITY DATA")
print("=" * 60)


# Rank weights
rank_weights = {
    "top_career_1": 5.0,
    "top_career_2": 4.0,
    "top_career_3": 3.0,
    "top_career_4": 2.0,
    "top_career_5": 1.0
}


# Create unique career list
all_careers = set()

for column in career_columns:

    all_careers.update(
        df_model[column]
        .dropna()
        .astype(str)
        .unique()
    )


all_careers = sorted(list(all_careers))

print("Unique careers:", len(all_careers))


# ============================================================
# 7. CREATE TRAINING ROWS
# ============================================================

training_X = []
training_y = []
training_weights = []


for i in range(len(df_model)):

    student_features = X.iloc[i].values

    for career in all_careers:

        score = 0.0

        # ----------------------------------------------------
        # Find career position
        # ----------------------------------------------------

        for column in career_columns:

            value = df_model.iloc[i][column]

            if pd.isna(value):
                continue

            if str(value) == career:

                score = rank_weights[column]
                break


        # ----------------------------------------------------
        # Positive examples
        # ----------------------------------------------------

        if score > 0:

            training_X.append(
                student_features
            )

            training_y.append(
                career
            )

            training_weights.append(
                score
            )


        # ----------------------------------------------------
        # Negative examples
        # ----------------------------------------------------
        # Add some careers that were NOT recommended.
        #
        # This helps the model learn the difference between
        # suitable and unsuitable careers.
        # ----------------------------------------------------

        else:

            # deterministic selection
            if (
                hash(
                    str(i) + career
                ) % 12 == 0
            ):

                training_X.append(
                    student_features
                )

                training_y.append(
                    "__NOT_RECOMMENDED__"
                )

                training_weights.append(
                    0.2
                )


training_X = np.array(training_X)

training_y = np.array(training_y)

training_weights = np.array(
    training_weights
)


print(
    "Training samples:",
    len(training_X)
)

print(
    "Training classes:",
    len(np.unique(training_y))
)


# ============================================================
# 8. TRAIN / TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test, w_train, w_test = (
    train_test_split(
        training_X,
        training_y,
        training_weights,
        test_size=0.20,
        random_state=42,
        stratify=training_y
    )
)


# ============================================================
# 9. TRAIN MODEL
# ============================================================

print()
print("=" * 60)
print("TRAINING IMPROVED RANKING MODEL")
print("=" * 60)


ranking_model = HistGradientBoostingClassifier(
    max_iter=600,
    learning_rate=0.035,
    max_leaf_nodes=31,
    min_samples_leaf=5,
    l2_regularization=1.5,
    early_stopping=True,
    random_state=42
)


ranking_model.fit(
    X_train,
    y_train,
    sample_weight=w_train
)


# ============================================================
# 10. TEST
# ============================================================

pred = ranking_model.predict(
    X_test
)


accuracy = accuracy_score(
    y_test,
    pred
)


print()
print("=" * 60)
print("MODEL RESULTS")
print("=" * 60)

print(
    f"Classification accuracy: "
    f"{accuracy * 100:.2f}%"
)


# ============================================================
# 11. SAVE MODEL
# ============================================================

ranking_package = {

    "model": ranking_model,

    "feature_columns": ranking_features,

    "career_list": all_careers,

    "rank_weights": rank_weights,

    "accuracy": accuracy
}


joblib.dump(
    ranking_package,
    "career_ranking_model.pkl"
)


print()
print("=" * 60)
print("RANKING MODEL SAVED")
print("=" * 60)

print(
    "File: career_ranking_model.pkl"
)

print(
    "Features:",
    len(ranking_features)
)

print(
    "Careers:",
    len(all_careers)
)

print(
    f"Accuracy: "
    f"{accuracy * 100:.2f}%"
)

print("=" * 60)
```
